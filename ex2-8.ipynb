```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

def read_data(fname):
    df = pd.read_csv(fname, sep=' ', header=None)
    df = df[df[0].isin([2, 3])]
    x = df.values[:,1:257]
    y = df.values[:,0] == 3
    return x, y

x_train, y_train = read_data('data/zip.train')
x_test, y_test = read_data('data/zip.test')

m = LinearRegression().fit(x_train, y_train)

y_train_pred = m.predict(x_train) >= 0.5
y_test_pred = m.predict(x_test) >= 0.5

train_acc = accuracy_score(y_train, y_train_pred)
test_acc = accuracy_score(y_test, y_test_pred)

print(f'{"method":<20}{"train acc":>12}{"test acc":>12}')
print(f'{"linear regression":<20}{train_acc:>12.4f}{test_acc:>12.4f}')

for k in [1, 3, 5, 7, 15]:
    m = KNeighborsClassifier(n_neighbors=k).fit(x_train, y_train)

    y_train_pred = m.predict(x_train)
    y_test_pred = m.predict(x_test)

    train_acc = accuracy_score(y_train, y_train_pred)
    test_acc = accuracy_score(y_test, y_test_pred)

    print(f'{f"{k}-NN":<20}{train_acc:>12.4f}{test_acc:>12.4f}')
```
